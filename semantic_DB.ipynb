```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
# 1. Load a lightweight, fast English embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Prepare sample English text data (Knowledge Base)
documents = [
    "Ahmed works as a cardiac surgeon at London Hospital.",               # ID: 0
    "Sara is a software engineer building artificial intelligence apps.", # ID: 1
    "Khaled is a pulmonologist treating asthma and breathing problems.",  # ID: 2
    "The National Library is located right in the city center."           # ID: 3
]

# 3. Convert text documents into float32 numerical vectors (Embeddings)
embeddings = model.encode(documents)
embeddings = np.array(embeddings).astype('float32')

# Get the vector dimension (shape[1] gives 384 for this model)
dimension = embeddings.shape[1]

# 4. Initialize FAISS index (Exact L2 Euclidean distance) and add vectors
index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

print(f"Successfully indexed {index.ntotal} documents!\n")

# 5. Run a semantic search query using different wording
query = "Which doctor can help with chest pain and coughing?"
query_vector = np.array(model.encode([query])).astype('float32')

# Retrieve top 2 closest matches
k = 2
distances, indices = index.search(query_vector, k)

# 6. Print the matched results
print(f"Query: '{query}'\n")
print("Top Semantic Matches:")
for i in range(k):
    doc_id = indices[0][i]       # Extract document ID from 2D result array
    distance = distances[0][i]   # Extract distance score (lower means closer meaning)
    print(f"- Match {i+1} (ID {doc_id}): \"{documents[doc_id]}\" | Distance: {distance:.4f}")
```
